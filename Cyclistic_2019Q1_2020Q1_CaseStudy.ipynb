## Introduction
This is a case study for the fictional company, Cyclistic. Cyclistic is a bike-sharing company with two types of memberships, casual riders and annual riders. Customers who purchase single-ride or day passes are categorized as casual riders while customers who purchase the annual pass are categorized as members or subscribers.
## Objective
The purpose of this case study is to identify how casual riders use Cyclistic bikes differently than annual members, in order to provide insights that can increase membership conversion.

**Key Findings**



*   In general, casual riders have longer duration of rides than that of members.
*   On the contrary, members have a greater number of rides per day than casual riders.
*   Casual riders use the program more on the weekends, Saturday and Sunday. This could lead to believe casual riders use the program for hobby or exercise reasons.
*   Program members use the program less on the weekends, Saturday and Sundays. This could mean that members of the program use it as a mode of transportation during the weekdays, Mon-Fri.




## Recommendations




1.  **Program Points and Rewards:** Since casual riders tend to have longer ride durations, offer an exclusive membership perk that allows points to be rewarded based on the duration/distance of rides. Allow the members to redeem accumulated points for free rides, membership discounts, etc. Send promotional items to all users as an incentive to not only convert casual riders but an incentive to increase the ride duration of existing members as well.
2.   **Commuter Pass:** Considering most casual riders don't use the program as much during weekdays, introduce a 'Commuter Pass' which is a discounted annual membership specifically for weekday usage (Monday-Friday). This targets casual riders who might use the service for commuting but are deterred by the full annual price, or who only need it for specific days. Highlight the cost savings compared to purchasing multiple single rides.
3.   **Rider Challenges:** Create some kind of gamified challenges that unlock rewards for free rides, membership discounts, etc. This leverages already exisiting usage patterns and provides an incentive for casual riders to purchase that subscription.
